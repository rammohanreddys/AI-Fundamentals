#### Module:
In Python, both modules and packages are tools used to organize and modularize your code, but they exist at different levels of the file structure hierarchy.

Think of a module as a single file, and a package as a folder full of those files.

##### 1. What is a Module?

A module is simply a single Python file ending in .py. It can contain variables, functions, classes, and runnable code. You use modules to break down large programs into manageable, reusable pieces.

- **File Structure:** A single file (e.g., mymodule.py).

- **How to use it:** You bring it into another file using the import statement.

```python
file: calculator.py (This is a module)

def add(x, y):
    return x + y

#* file: main.py
import calculator
print(calculator.add(5, 3))
```

#### Package:

##### 2. What is a Package?

A package is a collection of Python modules housed inside a directory (folder). It allows you to organize modules hierarchically using "dot notation" (e.g., package.module).

- File Structure: A directory containing .py files.

- The __init__.py file: Historically (and as a best practice), a package must contain a file named __init__.py. This file can be completely empty, but its presence tells Python, "Treat this directory as a Python package."



To use code from a package, you use the dot syntax:

#### File Operations:

Working with files is a fundamental skill in Python. Whether you are data mining, logging application events, or saving user settings, you will need to interact with the file system.

##### 1. The Core Concept: The open() Function

To interact with a file, you must first open it. Python achieves this using the built-in open() function, which creates a file object (stream).

**Syntax:**

file_object = open(file_name, mode)

##### File modes:

The mode argument tells Python what you intend to do with the file.

|Mode | Name | Description |
|------|-----|-------------|
| r | Read (Default)    | Opens a file for reading. Raises an error if the file doesn not exist.|
| w | Write             | Opens a file for writing. Overwrites the file if it exists, or creates a new one.|
| a | Append            | Opens a file for writing. Appends data to the end of the file without overwriting.|
| x | Exclusive Creation| Creates a new file. Raises a FileExistsError if it already exists. |
| b | Binary            | Opens the file in binary mode (e.g., 'rb', 'wb') for images, PDFs, etc.|
| + | Update            | Opens a file for both reading and writing (e.g., 'r+').|



##### Basic File Operations (The Manual Way):

In the traditional approach, you open the file, perform your operations, and must explicitly close it. If you forget to close it, the file remains locked in memory, which can lead to leaks or data corruption.

**Writing to a File:**

```python
file = open("example.txt", "w")
file.write("Hello World!\n")
file.write("This is a new line.")
file.close()  # Extremely important!
```

**Reading from a File** - There are three main ways to read a text file:

```python
file = open("example.txt", "r")

# 1. Read the entire file content as a single string
content = file.read() 

# 2. Read only the first 5 characters
# partial_content = file.read(5) 

# 3. Read line by line into a list
# lines = file.readlines() 

print(content)
file.close()
```

**The Modern Standard: Context Managers (with statement)**

In real-world Python, you should almost never manually close files. Instead, use the with statement. This is a context manager that guarantees the file will be closed automatically when the code block finishes executing—even if an error occurs inside the block.

```python
# The file is safely closed the moment you indent backward out of this block
with open("example.txt", "r") as file:
    content = file.read()
    print(content)
```

```python
## Read a file line by line
with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) ## strip() removes the newline character
```

```python
## Writing a file(Overwriting)

with open('example.txt','w') as file:
    file.write('Hello World!\n')
    file.write('this is a new line.')
```

```python
## Write a file(wwithout Overwriting)
with open('example.txt','a') as file:
    file.write("Append operation taking place!\n")
```

```python
### Writing a list of lines to a file
lines=['First line \n','Second line \n','Third line\n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
### Binary Files

# Writing to a binary file
data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
### Read the content froma  source text fiile and write to a destination text file
# Copying a text file
with open('example.txt', 'r') as source_file:
    content = source_file.read()

with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
#Read a text file and count the number of lines, words, and characters.
# Counting lines, words, and characters in a text file
def count_text_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        line_count = len(lines)
        word_count = sum(len(line.split()) for line in lines)
        char_count = sum(len(line) for line in lines)
    return line_count, word_count, char_count

file_path = 'example.txt'
lines, words, characters = count_text_file(file_path)
print(f'Lines: {lines}, Words: {words}, Characters: {characters}')
```

```python
### Writing and then reading a file

with open('example.txt','w+') as file:
    file.write("Hello world\n")
    file.write("This is a new line \n")

    ## Move the file cursor to the beginning
    file.seek(0)

    ## Read the content of the file
    content=file.read()
    print(content)
```

#### Exception Handling:

Exception handling in Python allows you to handle errors gracefully and take corrective actions without stopping the execution of the program. This lesson will cover the basics of exceptions, including how to use try, except, else, and finally blocks.

**What Are Exceptions?**

Exceptions are events that disrupt the normal flow of a program. They occur when an error is encountered during program execution. Common exceptions include:

- ZeroDivisionError: Dividing by zero.
- FileNotFoundError: File not found.
- ValueError: Invalid value.
- TypeError: Invalid type.

```python
a=b
```

```python
try:
    a=b
except NameError:
    print("value 'b' not assigned")
```

```python
10/0
```

```python
try:
    result=10/0
except ZeroDivisionError:
    print("denominator should not be zero")
except Exception as ex:
    print(ex)
finally:
    print("program execution complted")
```
