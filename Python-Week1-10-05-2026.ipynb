### Variables in Python:

In Python, a variable is essentially a reserved memory location to store values.

#### 1. Creating Variables:

x = 5          # x holds an integer
name = "Alice" # name holds a string

#### 2. Naming Rules (Syntax):

Python is strict about how you name your "boxes." If you break these rules, your code will crash:

- Must start with a letter or the underscore character (_).

- Cannot start with a number.

- Can only contain alpha-numeric characters and underscores (A-z, 0-9, and _).

- Case-sensitive: age, Age, and AGE are three different variables.

- Reserved Words: You cannot use Python "keywords" (like if, else, print, or while) as variable names.

#### 3. Dynamic Typing:

Python is dynamically typed. This means you don't have to specify what type of data the variable holds (like "this is a number"), and you can even change the type of data stored in a variable later on.

data = 10       # 'data' is an integer
data = "Hello"  # 'data' is now a string

#### 4. Multiple Assignment:

Python allows you to assign values to multiple variables in one line, which makes your code very clean:

#Assigning different values to multiple variables
a, b, c = 1, 2, "Three"

#Assigning the same value to multiple variables
x = y = z = 100

#### 5. Standard Data Types:

While you don't have to declare the type, every variable has a type based on the value it holds. You can check this using the type() function.

| str | z = "Hi" | Text wrapped in quotes. |

#### 6. Variable Scope:

- **Global Variables:** Created outside of a function and accessible by everyone.

- **Local Variables:** Created inside a function and can only be used within that function.

### Data Types in Pyhton:

Python has a rich and flexible type system. A “data type” defines what kind of value a variable can hold and what operations can be performed on it.
Python’s data types are more than just labels for values; they are objects. In Python, every variable is a reference to an object in memory. Understanding these "in deep" means looking at their **mutability**, **memory allocation**, and **internal structure**.

#### 1. What Are Data Types in Python?

In Python, everything is an object.

x = 10
print(type(x))

This means:
- 10 is an object
- its type is int

Python is:

- Dynamically typed → no need to declare types
- Strongly typed → Python does not automatically mix incompatible types

Example:

x = 5

x = "hello"

Same variable can hold different types.


```python
x = 5
print(x)
print(type(x))

x = "hello"
print(x)
print(type(x))
```

#### list vs tupple in packing & unpacking:

Packing occurs when you assign multiple values to a single variable.

**Tuples (Default):** Python defaults to tuples when you pack values without explicit brackets. This is often called "Implicit Tuple Packing."

**Lists:** You must explicitly use square brackets [] to pack values into a list.


##### Packing

```python
# Tuple Packing (Implicit)
t = 1, 2, 3 
# t is (1, 2, 3)
print(t)

# List Packing (Explicit)
l = [1, 2, 3] 
# l is [1, 2, 3]
print(l)
```

##### Unpacking:
Unpacking allows you to "distribute" the elements of a collection into separate variables. This works identically for both lists and tuples, provided the number of variables matches the number of elements.

```python
t = (10, 20)
a, b = t  # a=10, b=20
print(a)
print(b)

l = [30, 40]
c, d = l  # c=30, d=40
print(c)
print(d)
```

###### Extended Iterable Unpacking (The * Operator):
Python allows you to use the "splat" operator (*) to capture multiple items during unpacking. While you can unpack a tuple into a list or a list into a tuple, the collected part is always a list.

```python
# Unpacking a tuple, but collecting the "rest"
first, *middle, last = (1, 2, 3, 4, 5)

print(middle) 
# Output: [2, 3, 4]  <-- Note: This is a list even though the source was a tuple
```

#### Range (range): Efficient sequence generator.

Range() is used to generate a sequence of numbers.
It is commonly used with loops.

Basic Syntax:

range(start, stop, step)

```python
r = range(5)

print(r)
```

```python
r = range(5)

for i in range(5):
    print (i)
```

```python
# start and stop

for i in range(2, 6):
    print(i)
```

```python
# Positive Step
for i in range(1, 10, 2):
    print(i)
```

```python
# Negative Step
for i in range(10, 0, -2):
    print(i)
```

#### Convert Range to List:
range() itself is not a list.

```python
r = range(5)

print(list(r))
```

#### Nested Loops with Range:

```python
for i in range(3):
    for j in range(2):
        print(i, j)
```

#### Using enumerate() Instead of Range:

```python
names = ["A", "B", "C"]

for i in range(len(names)):
    print(i, names[i])
```

```python
for index, value in enumerate(names):
    print(index, value)
```
