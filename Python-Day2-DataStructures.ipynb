### Slicing: 

|List | Tuple | Set | Dictionary |
|------|-----|------|------------|
|List slicing allows you to extract a portion of a list using a specific syntax. <br> <br> **Basic Syntax:** <br><br> list[start : stop : step] <br> * start → index where slicing begins (included) <br> * stop → index where slicing ends (excluded) <br> * step → number of steps to move|Tuple slicing works exactly like list slicing. <br> The only difference: tuples are immutable (you cannot modify them). <br><br> **Basic Syntax:** <br><br> tuple[start : stop : step]|Set does NOT support slicing in Python. <br> Because sets are unordered collections: <br><br> * ❌ No index positions <br> * ❌ No fixed order of elements <br>*  ✅ Elements are stored randomly|Slicing is NOT supported in dictionaries. <br> Because dictionaries are: <br> <br> * ❌ Not index-based <br> * ✅ Accessed using keys, not positions <br> <br> 👉 Slicing needs index/position, but dictionaries use key-value pairs|
